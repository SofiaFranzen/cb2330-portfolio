Dot product: $\sum_{i=1}^{N} a_i b_i$, computes how much two vectors of numbers agree

```python
A_numbers = [1,2,3]
B_numbers = [1,1,2]

def dot_product(A, B):
  total = 0
  for a, b in zip(A, B):
    total = total + a*b

  return total

print(dot_product(A_numbers, B_numbers))
```

```python
import numpy as np

def dot_product_numpy(A, B):
  dotproduct = np.sum(np.array(A)*np.array(B))

  return dotproduct

print(dot_product_numpy(A_numbers, B_numbers))
```
